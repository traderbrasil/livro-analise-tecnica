```python
import re
import pandas as pd

NOTEBOOKS = [
    ('1', 'cap01_introducao.ipynb', 'O seu primeiro notebook'),
    ('2', 'cap02_dow.ipynb', 'Dow com Python'),
    ('3', 'cap03_graficos.ipynb', 'O mesmo preço de vários jeitos'),
    ('4', 'cap04_tendencia.ipynb', 'Tendência, suportes e gaps'),
    ('5', 'cap05_padroes.ipynb', 'O scanner de padrões gráficos'),
    ('6', 'cap06_fibonacci.ipynb', 'Fibonacci no preço e no tempo'),
    ('7', 'cap07_elliott.ipynb', 'As ondas de Elliott em Python'),
    ('8', 'cap08_indicadores.ipynb', 'Indicadores escritos à mão'),
    ('9', 'cap09_seguidores.ipynb', 'Seguidores de tendência'),
    ('9', 'cap09b_kama_tartarugas.ipynb', 'Kaufman e as tartarugas'),
    ('9', 'cap09c_dcatr.ipynb', 'O DC ATR de David Cox'),
    ('10', 'cap10_momentum.ipynb', 'Osciladores de momentum'),
    ('11', 'cap11_volume.ipynb', 'Volume, o combustível do mercado'),
    ('12', 'cap12_folego.ipynb', 'Fôlego de mercado (Market Breadth) na B3'),
    ('13', 'cap13_sentimento.ipynb', 'Sentimento, o termômetro do medo e da ganância'),
    ('14', 'cap14_ciclos.ipynb', 'Ciclos e sazonalidade'),
    ('15', 'cap15_gann.ipynb', 'Gann sem astrologia'),
    ('16', 'cap16_forca_relativa.ipynb', 'Força relativa'),
    ('17', 'cap17_argon.ipynb', 'A grade de nove (versão didática)'),
    ('18', 'cap18_intermercados.ipynb', 'Intermercados e liquidez'),
    ('19', 'cap19_cripto.ipynb', 'Criptoativos'),
    ('20', 'cap20_wyckoff.ipynb', 'Wyckoff'),
    ('21', 'cap21_smc.ipynb', 'SMC e manipulação'),
    ('22', 'cap22_profile.ipynb', 'Market Profile e Volume Profile'),
    ('23', 'cap23_price_action.ipynb', 'Price action, Al Brooks e Linda Raschke'),
    ('24', 'cap24_strat.ipynb', 'The Strat'),
    ('25', 'cap25_estrategias.ipynb', 'Estratégias clássicas'),
    ('22 a 25', 'futuros_b3.ipynb', 'WIN e WDO no 5 minutos'),
    ('26', 'cap26_python.ipynb', 'Python para o analista técnico'),
    ('27', 'cap27_backtest.ipynb', 'O backtest em 10 passos'),
    ('27', 'cap27_sistemas.ipynb', 'Sistemas, backtesting e aprendizado de máquina'),
    ('28', 'cap28_stops.ipynb', 'Stops, o seguro que mantém você no jogo'),
    ('29', 'cap29_risco.ipynb', 'Gestão quantitativa do risco, o tamanho da mão'),
    ('30', 'cap30_psicologia.ipynb', 'Psicologia do trader em números'),
    ('31', 'cap31_roteiro.ipynb', 'Juntando tudo, o Método 6C testado'),
    ('32', 'cap32_conclusao.ipynb', 'As regrinhas da 3ª edição postas à prova'),
    ('Anexo I', 'anexo1_candlesticks.ipynb', 'Os padrões de candlesticks postos à prova'),
    ('Anexo II', 'anexo2_perpetuo.ipynb', 'Emendando os vencimentos do contrato futuro'),
    ('Anexo III', 'anexo3_opcoes.ipynb', 'O rabo, o cachorro e a volatilidade'),
    ('Anexo IV', 'anexo4_mapa.ipynb', 'O mapa de estudo para o CNPI-T e o CMT'),
    ('Apêndice', 'apendice_murphy.ipynb', 'Gráficos de longo prazo e inflação: o caso brasileiro'),
    ('Anexo V', 'anexo5_hub.ipynb', 'O índice do hub: notebooks, apps e episódios por capítulo'),
]

APPS = [
    ("ARGON", "A grade de nove de Ralph Acampora num universo de ações, com o fôlego das casas de cima e de baixo",
     "16, 17, 31", "argonglobalstocks.com (também em flaviolemoscmt.com)"),
    ("Market Breadth", "Fôlego da B3 e de outros mercados; scanners de máximas e mínimas, Bandas de Bollinger,"
     " cruzamentos de médias, pivôs, IFR e ADX; aba STRAT/TFC", "3, 12, 24, 31", "flaviolemoscmt.com"),
    ("Fear & Greed", "Índice de medo e ganância do S&P 500 e do Ibovespa, com 10 categorias", "13, 31",
     "flaviolemoscmt.com"),
    ("Seasonality", "Sazonalidade de um ativo mês a mês e screener de janelas sazonais", "14, 31", "flaviolemoscmt.com"),
    ("Momentum Dashboard", "Ranking de momentum (Clenow), scanner de IFR e fôlego acima das médias", "4, 5, 8, 9, 10",
     "flaviolemoscmt.com"),
]

EPISODIOS = [
    ('Steve Nison', 'Steve Nison: O Homem que Trouxe os Candlesticks para o Ocidente (Aula Magna)', 'fev. 2024', 'SSILYvMEOO4', '3, Anexo I'),
    ('Tom DeMark', 'Tom DeMark: o Segredo do Market Timing do Consultor de George Soros e Steve Cohen', 'jan. 2024', 'NRQ6anOsoqM', '1, 4, 13, 19'),
    ('Tyler Wood', 'Por Que a Certificação CMT Vai Mudar Sua Carreira? O CEO da Associação Explica', 'dez. 2023', 'np0MfJW7uLs', '1, Anexo IV'),
    ('Greg Morris', 'Pergunta de US$ 5,5 bilhões: O que um trader de elite sabe que você não sabe?', 'nov. 2023', 'ptwsPlexspQ', 'só no hub'),
    ('David Keller', 'David Keller, CMT (StockCharts): A Visão do Especialista em Análise Comportamental e Gráficos', 'nov. 2023', 'ZSqyNT7nnIo', '2, 30, 31'),
    ('Ivan Scherman', 'Desvendando os Segredos do Campeão Mundial de Futuros: Ivan Scherman', 'out. 2023', 'yZUUlB4HF1I', '1, 27, 29'),
    ('Bhavya Gehlot', 'Da Índia ao Mercado Global: Bhavya Gehlot Revela Como Operar Price Action', 'out. 2023', 'LgfStC-bBj4', 'só no hub'),
    ('John e Zoe Bollinger', 'Bandas de Bollinger: o Erro nº 1 que Todo Trader Comete', 'set. 2023', 'QkOFDk8uVi4', '1, 8, 9, 20'),
    ('Alex Spiroglou', 'Você é um S.M.A.R.T trader? O Criador do MACD-V explica', 'out. 2024', 'd6U1Js8bn98', '8, 10'),
    ('Jake Bernstein', 'Jake Bernstein: “70% é Psicologia, 30% é Estratégia”', 'set. 2024', 'q8OxEqSWck0', '9, 14, 29, 30, 31'),
    ('Buff Dormeier', 'Análise de Volume no Trading: Como Operar com a “Força” e ser o Yoda', 'ago. 2024', 'QWLiPu5ebxs', '2, 11'),
    ('Brian Shannon', 'Criador do VWAP Ancorado Revela Tudo: Entradas, Saídas e Múltiplos Tempos Gráficos', 'jul. 2024', 'arDR21c1rpE', '3, 5, 11, 22, 28'),
    ('John Kosar', 'Dez Anos no Pregão da CME: John Kosar Revela o Que Realmente Move os Preços', 'maio 2024', 'K-anCrPWWqU', 'só no hub'),
    ('Aksel Kibar', 'Aksel Kibar, CMT: O Mestre dos Gráficos Clássicos e Gestão de Fundos em Abu Dhabi', 'abr. 2024', 'fVucWfg8QNg', '4, 5'),
    ('Márcio Kieling', 'Ele trocou a Globo pelo Gráfico: O Setup MK e o Pior Trade!', 'fev. 2026', 'yQfzT7DfFNI', '3, 25'),
    ('Enrico Cozzolino', 'Filho Rico de Pai Pobre: Por Que Você Não Enriqueça', 'fev. 2026', 'NTtZmD1IEro', 'só no hub'),
    ('Giba Coelho', 'PARE DE SOFRER NO DAY TRADE: O “Gibex Sossegado” de Giba Coelho (XP) Revelado', 'fev. 2026', '2DuLPxgRNUs', '25'),
    ('Damanick Dantes', 'Bitcoin ou Ouro? A Estratégia de Rotação que Gestores Institucionais Usam', 'jan. 2026', 'V1981G00dL8', 'só no hub'),
    ('Jyoti Budhia', 'Jyoti Budhia, a “Rainha das Opções”: Como ela sobrevive e prospera no mercado de ações', 'jan. 2026', '_X3cyOLUU3w', 'Anexo III'),
    ('Jamie Coutts', 'A verdade sobre Bitcoin e inflação que Jamie Coutts, CMT acabou de revelar', 'jun. 2026', 'U9fEeOkCJ3o', '18, 19'),
    ('Razan Hilal', 'Bitcoin US$ 200 mil? O plano macro de criptomoedas de Razan Hilal, CMT', 'dez. 2025', 'as7aB10WB48', '7, 14, 18'),
    ('Jim Bianco', 'O Mercado de Bonds Quebrou: 60/40 Está Morto, “Boomer Bitcoin” e o Fed Fora de Controle', 'dez. 2025', '5-G2xISbhbU', '18'),
    ('Michael Nauss', 'Ele Perdeu Uma Fortuna Ignorando o Próprio Sistema, e Mudou Tudo', 'maio 2026', 'S2ynKL_Do_Y', '30'),
    ('Ralph Acampora', 'Análise Técnica Está Morta? O Padrinho Responde', 'dez. 2025', 'XgSTi7exrys', '1, 2, 3, 16, 17, Anexo I'),
    ('Sowmi Krishnamurthy', 'Não olhe só o VIX! A “Trindade da Volatilidade” que Prevê os Crashes', 'abr. 2026', 'BF2_9eEwD5k', 'só no hub'),
    ('Foram Chheda', 'Ouro vs S&P 500: Onde o Dinheiro Inteligente Está Escondido?', 'mar. 2026', 'uCyycJ9work', '16, 18, 20'),
    ('Tom Hougaard', 'O Melhor Perdedor Vence: Tom Hougaard e a Psicologia do Mercado no Day Trade', 'mar. 2026', '0YWf9ipbYP8', '28, 29, 30'),
    ('Walter Deemer', 'Quando a Hora de Comprar Chegar, Você Não Vai Querer', 'mar. 2026', 'CcjKhzCNdjM', '4, 14'),
    ('Craig Johnson', 'Craig Johnson, CMT: Ele Aquece a Casa com Bitcoin', 'mar. 2026', 'SyfEVzt8xZk', '16, 26'),
    ('Kelly Corbiere', 'Psicologia do Trader: Erro do Trader Individual', 'fev. 2026', 'vTQSn7ZSn_A', '1'),
    ('Carolyn Boroden', 'Carolyn Boroden: A Rainha de Fibonacci Revela tudo sobre Clusters, Simetria e Timing', 'abr. 2026', 'kA6U4XOXnH8', '1, 6'),
    ('Al Brooks', 'Al Brooks: A Lenda do Price Action Revela a “Bíblia” do Day Trade', 'jan. 2026', 'DEm9TTKxNH8', '4, 21, 23'),
    ('Haroldo MacDowell', 'A Geometria Secreta do Mercado: Gann, Elliott e o “Bruxo”', 'jun. 2026', 'xIAx3Pza4zE', '15'),
    ('Kavita Agrawal', 'Como Fazer Swing Trade com Força Relativa (RSQM)', 'maio 2026', 'aiQBFxESpIM', '16'),
    ('Jason Shapiro', 'A regra de ouro de Jason Shapiro', 'maio 2026', 'd4bFigmawII', '13'),
    ('John Murphy (palestra de 2011)', 'John Murphy In Memoriam: A Masterclass Perdida sobre Análise Intermercados (2011)', 'abr. 2026', 'WsYx2Dz1Mlg', '16, 18'),
    ('Lucas Costa', 'De Viral no Twitter a Head Análise Técnica Banco BTG', 'set. 2026', '5UojC9blcGM', 'só no hub'),
    ('Steve Hochberg', 'Ondas de Elliott e Socionomia: O Fim do Bull Market', 'set. 2026', 'GnvIEkV9Mds', '7'),
    ('David Tonaszuck', 'A Transição do Bloomberg para o Microsoft Office', 'set. 2026', '02cw_iZrt9g', 'só no hub'),
    ('Brett Steenbarger', 'Você Está Suprimindo Emoções? O Erro Fatal', 'ago. 2026', 'ZG52JawNS6M', '30'),
    ('Katie Stockton', 'Sistemático vs. Emocional: O Método de Katie Stockton', 'jul. 2026', 'oGR71ZNyu2Q', '16'),
    ('Gina Martin Adams', 'Por que as Mag-7 Estão em Perigo: O Alerta do HHI', 'jul. 2026', 'h898Jin_0xc', '12'),
    ('Marcio Ferracini', 'O Fim da Valuation? A Fórmula Secreta', 'jun. 2026', 'c1hkppcAzS0', '1'),
    ('Raphael Figueredo', 'Do Hip Hop a Wall Street: A Fórmula Secreta de Rafi', 'jun. 2026', 'XQB-a9uF4zM', '10'),
]


def mostra(*partes):
    """print com vírgula decimal e ponto de milhar (padrão brasileiro)."""
    texto = " ".join(str(p) for p in partes)
    texto = re.sub(r"(?<=\d)\.(?=\d)", ",", texto)
    print(re.sub(r"(?<=\d)_(?=\d)", ".", texto))


def tem(lista, cap):
    """True se o capítulo (ex.: '16' ou 'Anexo I') está na lista '16, 17, 31'."""
    return cap in [c.strip() for c in lista.split(",")]
```

## Parte 1: os notebooks do livro

```python
NB = pd.DataFrame(NOTEBOOKS, columns=["capítulo", "notebook", "o que faz"])
mostra(f"Parte 1. {len(NB)} notebooks em {NB.loc[~NB['capítulo'].str.contains(' a '), 'capítulo'].nunique()} capítulos e anexos")
for _, l in NB.iterrows():
    mostra(f"   {l['capítulo']:>8}  {l['notebook']:<30}  {l['o que faz'][:70]}")
```

## Parte 2: os apps do autor

```python
AP = pd.DataFrame(APPS, columns=["app", "o que faz", "capítulos", "endereço"])
mostra(f"\nParte 2. {len(AP)} ferramentas do autor")
for _, l in AP.iterrows():
    mostra(f"   {l['app']:<46} capítulos {l['capítulos']:<14} {l['endereço']}")
```

## Parte 3: os episódios do Traders of the World

```python
EP = pd.DataFrame(EPISODIOS, columns=["convidado", "episódio", "data", "id", "capítulos"])
EP["link"] = "https://www.youtube.com/watch?v=" + EP["id"]
citados = EP[EP["capítulos"] != "só no hub"]
mostra(f"\nParte 3. {len(EP)} episódios no catálogo; {len(citados)} citados no livro")
for _, l in citados.iterrows():
    mostra(f"   {l['convidado']:<24} {l['data']:<10} capítulos {l['capítulos']}")
```

## Parte 4: o índice de um capítulo

```python
def indice(cap):
    """Tudo o que existe fora do papel para um capítulo: notebooks, apps e episódios."""
    mostra(f"\nÍndice do {'Capítulo ' + cap if not cap.startswith(('Anexo', 'Apêndice')) else cap}")
    for _, l in NB[NB["capítulo"] == cap].iterrows():
        mostra(f"   notebook: {l['notebook']} ({l['o que faz']})")
    for _, l in AP[AP["capítulos"].map(lambda s: tem(s, cap))].iterrows():
        mostra(f"   app: {l['app']} ({l['endereço']})")
    for _, l in EP[EP["capítulos"].map(lambda s: tem(s, cap))].iterrows():
        mostra(f"   episódio: {l['convidado']}, “{l['episódio']}” ({l['data']}): {l['link']}")


mostra("\nParte 4. O índice de um capítulo (troque o número)")
indice("16")
```

```python
# Troque o capítulo e veja tudo o que existe fora do papel para ele
indice("27")
indice("Anexo I")
```

## Parte 5: o índice inteiro numa planilha

```python
linhas = []
for _, l in NB.iterrows():
    linhas.append({"capítulo": l["capítulo"], "tipo": "notebook", "nome": l["notebook"], "detalhe": l["o que faz"]})
for _, l in AP.iterrows():
    for c in l["capítulos"].split(","):
        linhas.append({"capítulo": c.strip(), "tipo": "app", "nome": l["app"], "detalhe": l["endereço"]})
for _, l in citados.iterrows():
    for c in l["capítulos"].split(","):
        linhas.append({"capítulo": c.strip(), "tipo": "episódio", "nome": f"{l['convidado']}: {l['episódio']}",
                       "detalhe": l["link"]})
IDX = pd.DataFrame(linhas)
IDX.to_csv("indice_hub.csv", index=False, encoding="utf-8-sig")
mostra(f"\nParte 5. indice_hub.csv gravado: {len(IDX)} linhas ({(IDX['tipo'] == 'notebook').sum()} notebooks,"
       f" {(IDX['tipo'] == 'app').sum()} apps por capítulo, {(IDX['tipo'] == 'episódio').sum()} episódios por capítulo)")
```
